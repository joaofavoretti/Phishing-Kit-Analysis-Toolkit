```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
import numpy as np
```

```python
sentences = [line.strip() for line in open('./instruction_sentences.txt', 'r').readlines()]
```

```python
sentences_data = []

for i, sentence in enumerate(sentences):
    sentences_data.append(TaggedDocument(sentence.split(), [i]))
```

```python
model = Doc2Vec(vector_size=128, window=32, min_count=1, workers=4, epochs=40, dm=0, dbow_words=1)
model.build_vocab(sentences_data)
model.train(sentences_data, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
X = np.array([model.dv[tagged_doc.tags[0]] for tagged_doc in sentences_data])
y = np.array([tagged_doc.tags[0] for tagged_doc in sentences_data])
```

```python
type(X)
```

```python
with open('vectors_d2v.tsv', 'w') as f:
    for v in X:
        f.write('\t'.join(map(str, v)) + '\n')
```

```python
with open('metadata_d2v.tsv', 'w') as f:
    for i in y:
        f.write(str(i) + '\n')
```
