```python
from sentence_transformers import SentenceTransformer
```

```python
sentences = [line.strip() for line in open('./instruction_sentences.txt', 'r').readlines()]
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentence_embeddings = model.encode(sentences)
print(sentence_embeddings.shape)
print(sentence_embeddings)

sentence_similarities = model.similarity(sentence_embeddings, sentence_embeddings)
print(sentence_similarities)
```

```python
with open('vectors.tsv', 'w') as f:
    for sentence in sentence_embeddings:
        f.write('\t'.join([str(x) for x in sentence]) + '\n')
```

```python
with open('metadata.tsv', 'w') as f:
    for idx in range(len(sentences)):
        f.write(f'{idx}\n')
```

```python
idx = 550

with open(f'sentence{idx}.txt', 'w') as f:
    for word in sentences[idx].split():
        f.write(word + '\n')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(font_scale=1.2)
plt.figure(figsize=(10, 10))
sns.heatmap(sentence_similarities, annot=True, fmt=".2f", xticklabels=sentences, yticklabels=sentences, cmap='coolwarm')
```
